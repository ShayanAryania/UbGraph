```python
import pandas as pd
import matplotlib.pyplot as plt

day9_key = pd.DataFrame({
    "Model": [
        "ESM2-3B Fusion XGB",
        "ESM2-3B Seq XGB",
        "ESM2-650M Seq XGB",
        "ESM2-650M Fusion XGB",
        "ProtBERT-BFD Fusion XGB",
        "ESM2-3B Fusion RF"
    ],
    "PR_AUC": [0.8463, 0.8421, 0.8421, 0.8408, 0.8355, 0.8368],
    "ROC_AUC": [0.7594, None, None, None, None, None],
    "F1": [0.7562, None, None, None, None, None]
})

day10_fusion = pd.DataFrame({
    "Fusion": [
        "ESM650 + ProtBERT-BFD",
        "ESM3B + ProtBERT-BFD",
        "ESM650 + ESM3B",
        "ESM650 + ESM3B + ProtBERT-BFD"
    ],
    "Features": [9408, 14528, 15552, 19744],
    "PR_AUC": [0.8430, 0.8477, 0.8476, 0.8500],
    "ROC_AUC": [0.7566, 0.7617, 0.7640, 0.7659]
})

best_compare = pd.DataFrame({
    "Model": [
        "Best single-model\nESM2-3B Fusion XGB",
        "Best cross-model fusion\nESM650 + ESM3B + ProtBERT-BFD XGB"
    ],
    "ROC_AUC": [0.7594, 0.7659],
    "PR_AUC": [0.8463, 0.8500],
    "F1": [0.7562, 0.7677],
    "Brier": [0.2154, 0.2117]
})

improvement = pd.DataFrame({
    "Metric": ["ROC-AUC", "PR-AUC", "F1"],
    "Improvement": [0.0065, 0.0037, 0.0115]
})

# Plot 1
day9_plot = day9_key.sort_values("PR_AUC", ascending=True)

plt.figure(figsize=(9, 5))
plt.barh(day9_plot["Model"], day9_plot["PR_AUC"])
plt.xlabel("PR-AUC")
plt.title("Day 9 — Key baseline model comparison by PR-AUC")
plt.xlim(0.82, 0.855)

for i, v in enumerate(day9_plot["PR_AUC"]):
    plt.text(v + 0.0005, i, f"{v:.4f}", va="center")

plt.tight_layout()
plt.savefig("e3uls_day9_key_baseline_pr_auc.png", dpi=300, bbox_inches="tight")
plt.show()

# Plot 2
fusion_plot = day10_fusion.sort_values("PR_AUC", ascending=True)

plt.figure(figsize=(9, 5))
plt.barh(fusion_plot["Fusion"], fusion_plot["PR_AUC"])
plt.xlabel("PR-AUC")
plt.title("Day 10 — Cross-model fusion ranking by PR-AUC")
plt.xlim(0.835, 0.855)

for i, v in enumerate(fusion_plot["PR_AUC"]):
    plt.text(v + 0.0004, i, f"{v:.4f}", va="center")

plt.tight_layout()
plt.savefig("e3uls_day10_cross_model_fusion_pr_auc.png", dpi=300, bbox_inches="tight")
plt.show()

# Plot 3
metrics_to_plot = ["ROC_AUC", "PR_AUC", "F1"]
compare_long = best_compare.melt(
    id_vars="Model",
    value_vars=metrics_to_plot,
    var_name="Metric",
    value_name="Score"
)

pivot = compare_long.pivot(index="Metric", columns="Model", values="Score")
ax = pivot.plot(kind="bar", figsize=(9, 5))

plt.ylabel("Score")
plt.title("Best single-model baseline vs best cross-model fusion")
plt.xticks(rotation=0)
plt.ylim(0.73, 0.87)

for container in ax.containers:
    ax.bar_label(container, fmt="%.4f", padding=3)

plt.tight_layout()
plt.savefig("e3uls_day9_vs_day10_best_model_metrics.png", dpi=300, bbox_inches="tight")
plt.show()

# Plot 4
plt.figure(figsize=(7, 5))
plt.bar(improvement["Metric"], improvement["Improvement"])
plt.ylabel("Absolute improvement")
plt.title("Improvement from best Day 9 model to best Day 10 fusion")

for i, v in enumerate(improvement["Improvement"]):
    plt.text(i, v + 0.0002, f"+{v:.4f}", ha="center")

plt.tight_layout()
plt.savefig("e3uls_day10_improvement_over_day9.png", dpi=300, bbox_inches="tight")
plt.show()

# Plot 5
plt.figure(figsize=(8, 5))
plt.scatter(day10_fusion["Features"], day10_fusion["PR_AUC"])
plt.xlabel("Number of features")
plt.ylabel("PR-AUC")
plt.title("Day 10 — Feature dimensionality vs PR-AUC")

for _, row in day10_fusion.iterrows():
    plt.text(row["Features"] + 150, row["PR_AUC"], row["Fusion"], fontsize=8)

plt.tight_layout()
plt.savefig("e3uls_day10_features_vs_pr_auc.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

neural_losses = pd.DataFrame({
    "Model": ["BCE", "Weighted BCE", "Focal Loss", "Hybrid Focal + Ranking"],
    "ROC_AUC": [0.763, 0.763, 0.773, 0.770],
    "PR_AUC": [0.844, 0.838, 0.846, 0.843]
})

pair_models = pd.DataFrame({
    "Model": [
        "Best XGBoost\nCross-model Fusion",
        "Neural Pair Scorer\nFocal Loss",
        "Soft Voting\nXGB + NN",
        "Stacking\nXGB + NN"
    ],
    "ROC_AUC": [0.766, 0.773, 0.782, 0.782],
    "PR_AUC": [0.850, 0.846, 0.860, 0.859],
    "F1": [None, None, 0.773, None]
})

gnn_models = pd.DataFrame({
    "Model": [
        "GraphSAGE",
        "GAT",
        "HeteroGraphSAGE",
        "HeteroGAT",
        "Edge-Aware HeteroGNN",
        "HGT",
        "Hetero Ensemble\nHGT+EdgeAware+HeteroSAGE",
        "Mega Stacking"
    ],
    "ROC_AUC": [0.8218, 0.7130, 0.8458, 0.8191, 0.8554, 0.8613, 0.8716, 0.8682],
    "PR_AUC": [0.8783, 0.7915, 0.9005, 0.8710, 0.9062, 0.9103, 0.9195, 0.9162]
})

final_metrics = pd.DataFrame({
    "Metric": ["ROC-AUC", "PR-AUC", "F1", "Accuracy", "Precision", "Recall", "Brier"],
    "Mean": [0.8716, 0.9195, 0.8327, 0.7955, 0.8346, 0.8319, 0.1490],
    "Std": [0.0192, 0.0160, 0.0239, 0.0218, 0.0308, 0.0354, 0.0162]
})

progression = pd.DataFrame({
    "Stage": [
        "Best XGB\nPair Fusion",
        "Best NN\nFocal Loss",
        "XGB+NN\nSoft Voting",
        "GraphSAGE",
        "HeteroGraphSAGE",
        "Edge-Aware\nHeteroGNN",
        "HGT",
        "Final Hetero\nEnsemble"
    ],
    "ROC_AUC": [0.766, 0.773, 0.782, 0.8218, 0.8458, 0.8554, 0.8613, 0.8716],
    "PR_AUC": [0.850, 0.846, 0.860, 0.8783, 0.9005, 0.9062, 0.9103, 0.9195]
})

graph_summary = pd.DataFrame({
    "Edge type": ["Enzyme-Substrate", "PPI", "Co-localization"],
    "Count": [6196, 60723, 193159]
})

# 1. Neural loss comparison
loss_long = neural_losses.melt(
    id_vars="Model",
    value_vars=["ROC_AUC", "PR_AUC"],
    var_name="Metric",
    value_name="Score"
)

pivot_loss = loss_long.pivot(index="Model", columns="Metric", values="Score").loc[neural_losses["Model"]]

ax = pivot_loss.plot(kind="bar", figsize=(9, 5))
plt.ylabel("Score")
plt.title("Day 11 — Neural Pair Scorer loss comparison")
plt.xticks(rotation=20, ha="right")
plt.ylim(0.82, 0.88)

for container in ax.containers:
    ax.bar_label(container, fmt="%.3f", padding=3)

plt.tight_layout()
plt.savefig("e3uls_day11_neural_loss_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# 2. Pair-level models
pair_plot = pair_models.sort_values("PR_AUC", ascending=True)

plt.figure(figsize=(9, 5))
plt.barh(pair_plot["Model"], pair_plot["PR_AUC"])
plt.xlabel("PR-AUC")
plt.title("Pair-level models before graph learning")
plt.xlim(0.835, 0.865)

for i, v in enumerate(pair_plot["PR_AUC"]):
    plt.text(v + 0.0004, i, f"{v:.3f}", va="center")

plt.tight_layout()
plt.savefig("e3uls_day11_pair_level_models_pr_auc.png", dpi=300, bbox_inches="tight")
plt.show()

# 3. GNN comparison
gnn_plot = gnn_models.sort_values("PR_AUC", ascending=True)

plt.figure(figsize=(10, 6))
plt.barh(gnn_plot["Model"], gnn_plot["PR_AUC"])
plt.xlabel("PR-AUC")
plt.title("Day 12 — Graph learning model comparison")
plt.xlim(0.75, 0.93)

for i, v in enumerate(gnn_plot["PR_AUC"]):
    plt.text(v + 0.002, i, f"{v:.4f}", va="center")

plt.tight_layout()
plt.savefig("e3uls_day12_gnn_model_pr_auc.png", dpi=300, bbox_inches="tight")
plt.show()

# 4. Performance progression
plt.figure(figsize=(11, 5))
plt.plot(progression["Stage"], progression["PR_AUC"], marker="o", label="PR-AUC")
plt.plot(progression["Stage"], progression["ROC_AUC"], marker="o", label="ROC-AUC")
plt.ylabel("Score")
plt.title("Performance progression from pair models to graph-aware ensemble")
plt.xticks(rotation=25, ha="right")
plt.ylim(0.74, 0.94)
plt.legend()

for i, row in progression.iterrows():
    plt.text(i, row["PR_AUC"] + 0.004, f'{row["PR_AUC"]:.3f}', ha="center", fontsize=8)

plt.tight_layout()
plt.savefig("e3uls_day12_model_progression.png", dpi=300, bbox_inches="tight")
plt.show()

# 5. Final model metrics with error bars
plt.figure(figsize=(9, 5))
plt.bar(final_metrics["Metric"], final_metrics["Mean"], yerr=final_metrics["Std"], capsize=4)
plt.ylabel("Score")
plt.title("Final best model: MegaWeighted Heterogeneous Ensemble")
plt.xticks(rotation=20, ha="right")
plt.ylim(0, 1.0)

for i, row in final_metrics.iterrows():
    plt.text(i, row["Mean"] + row["Std"] + 0.015, f'{row["Mean"]:.4f}', ha="center", fontsize=8)

plt.tight_layout()
plt.savefig("e3uls_day12_final_model_metrics_errorbars.png", dpi=300, bbox_inches="tight")
plt.show()

# 6. Graph edge composition
plt.figure(figsize=(8, 5))
plt.bar(graph_summary["Edge type"], graph_summary["Count"])
plt.ylabel("Number of edges")
plt.title("Day 12 — Biological graph edge composition")

for i, row in graph_summary.iterrows():
    plt.text(i, row["Count"] + 4000, f'{row["Count"]:,}', ha="center")

plt.tight_layout()
plt.savefig("e3uls_day12_graph_edge_composition.png", dpi=300, bbox_inches="tight")
plt.show()
```
